```python
import pandas as pd 
import os 

texts = []
filenames = []
for filename in os.listdir('data'):
    if filename.endswith('txt'):
        filenames.append(filename)
        texts.append(open('data/' + filename, 'r', encoding='utf-8').read())

df = pd.DataFrame({
    'Filename': filenames, 
    'Texts': texts
})

df.to_csv('annotations.csv', index=False)

df
        
```
